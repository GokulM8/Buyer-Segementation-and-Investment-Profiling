# Preprocessing

Clean the source tables, standardize column names, encode categorical fields, and prepare model-ready features.

```python
# Feature engineering: build client-level features and property features
from pathlib import Path
import sys
import pandas as pd

# ensure project root is on sys.path so `src` is importable when notebook is executed from notebooks/
proj_root = Path('..').resolve()
if str(proj_root) not in sys.path:
    sys.path.insert(0, str(proj_root))

from src.feature_engineering import build_client_feature_table, engineer_property_features
from src.utils import save_dataframe

# resolve data paths relative to project root (not notebooks/)
RAW = proj_root / 'data' / 'raw'
PROC = proj_root / 'data' / 'processed'
PROC.mkdir(parents=True, exist_ok=True)

# load CSVs directly from project data/raw
clients_fp = RAW / 'clients.csv'
properties_fp = RAW / 'properties.csv'
if not clients_fp.exists() or not properties_fp.exists():
    raise FileNotFoundError('Expected CSVs in data/raw (clients.csv, properties.csv)')

clients = pd.read_csv(clients_fp)
properties = pd.read_csv(properties_fp)

clients_features = build_client_feature_table(clients, properties)
properties_features = engineer_property_features(properties)

# preview
display(clients_features.head())
display(properties_features.head())

# save processed outputs
save_dataframe(clients_features, PROC / 'clients_features.csv')
save_dataframe(properties_features, PROC / 'properties_features.csv')

print('Saved processed datasets to', PROC)
```
